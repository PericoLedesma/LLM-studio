```python
# USE TOOL
async with client:
    result = await client.call_tool("greet", {"name": "Ford"})

print("\n🔧 Tool OUTPUT:")
print("=" * 50)
print(f"✓ Success: {not result.is_error}")
print(f"📄 Data: {result.data}")

if result.content:
    print(f"\n📝 Content:")
    for item in result.content:
        if hasattr(item, 'text'):
            print(f"   {item.text}")

if result.structured_content:
    print(f"\n📊 Structured: {result.structured_content}")

print("=" * 50)
```
